# Classificacao de Novos Dados

```python
import pandas as pd
import pickle

with open("../models/modelo_decision_tree.pkl", 'rb') as f:
    modelo = pickle.load(f)
```

## Classificacao dos Dados Inseridos pelo Usuario
pede os sete atributos do cliente, monta um DataFrame com as mesmas colunas usadas no treino e mostra o canal previsto pelo modelo. O processo se repete enquanto o usuário quiser.

```python
colunas = ['Region', 'Fresh', 'Milk', 'Grocery', 'Frozen', 'Detergents_Paper', 'Delicatessen']

continuar = 's'

while continuar == 's':
    print("dados do cliente:")
    region = int(input("Region (0 = Lisbon, 1 = Oporto, 2 = Other): "))
    fresh = float(input("Fresh: "))
    milk = float(input("Milk: "))
    grocery = float(input("Grocery: "))
    frozen = float(input("Frozen: "))
    detergents = float(input("Detergents_Paper: "))
    delicatessen = float(input("Delicatessen: "))

    # precisa ter os mesmos nomes de coluna
    novo_cliente = pd.DataFrame([[region, fresh, milk, grocery, frozen, detergents, delicatessen]], columns=colunas)

    resultado = modelo.predict(novo_cliente)

    if resultado[0] == 0:
        print("\ncanal de vendas previsto: HoReCa")
    else:
        print("\ncanal de vendas previsto: Retail")

    continuar = input("\nclassificar outro cliente? (s/n): ").lower()

print("saiu")
```
