# Analise Exploratoria de Dados (EDA)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("seaborn-v0_8")

df = pd.read_csv("../data/processed/wholesale_tratado.csv")
df.head()
```

## Estatisticas Descritivas
apresenta medidas como media, desvio padrao, minimo e maximo dos gastos, para entender a escala dos dados

```python
df.describe()
```

## Quantidade de Amostras por Canal
verifica quantos clientes existem em cada classe (0 e 1), para saber se as classes estao balanceadas

```python
df['Channel'].value_counts()
```

## Gasto Medio por Categoria em Cada Canal
compara o gasto medio de cada categoria, mostrando os atributos que mais os diferenciam

```python
gastos = ['Fresh', 'Milk', 'Grocery', 'Frozen', 'Detergents_Paper', 'Delicatessen']

media_canal = df.groupby('Channel')[gastos].mean()
media_canal.index = ['HoReCa', 'Retail']

media_canal.T.plot(kind='bar', figsize=(10,5))
plt.title("gasto medio por categoria")
plt.xlabel("categoria")
plt.ylabel("gasto anual")
plt.xticks(rotation=0)
plt.show()
```

## Distribuicao dos Gastos (Histograma)
mostra a distribuicao de gastos em cada categoria, para identificar assimetrias e outliers

```python
df[gastos].hist(figsize=(12,8), bins=30)
plt.suptitle("distribuicao dos gastos")
plt.show()
```

## Correlaçao entre os Atributos
gera um heatmap com a correlacao entre todas as colunas, para ver quais atributos tem mais relacao com o canal de vendas

```python
plt.figure(figsize=(9,6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("correlacao entre os atributos")
plt.show()
```
