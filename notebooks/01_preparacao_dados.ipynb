## Leitura dos Dados
carrega o arquivo

```python
import pandas as pd

df = pd.read_csv("../data/raw/wholesale.csv")
df.head()
```

## Verificacao dos Tipos das Colunas
Confere o tipo de cada coluna

```python
df.info()
```

## Conversao dos Valores
troca os textos pelos numeros definidos, usando a funcao `replace`


```python
# o modelo so aceita numeros
df['Channel'] = df['Channel'].replace({'HoReCa': 0, 'Retail': 1}).astype('int64')
df['Region'] = df['Region'].replace({'Lisbon': 0, 'Oporto': 1, 'Other': 2}).astype('int64')

df.head()
```

```python
df.dtypes
```

## Reordenação das Colunas
coloca a coluna Channel no final, pois ela e o rotulo de classe que o modelo vai prever


Para isso usamos a função `reindex` com o atributo `columns`

```python
nova_ordem = ['Region', 'Fresh', 'Milk', 'Grocery', 'Frozen', 'Detergents_Paper', 'Delicatessen', 'Channel']
df = df.reindex(columns=nova_ordem)

df.head()
```

## Salvamento do Arquivo Tratado
exporta os dados tratados para usar nas proximas etapas

```python
output_path = "../data/processed/wholesale_tratado.csv"
df.to_csv(output_path, index=False)
```
